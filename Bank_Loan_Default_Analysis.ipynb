```python
#rename columns to snake_case
#'Default' is a reserved word in SQL, so it becomes default_flag
df.columns = ['loan_id', 'age', 'income', 'loan_amount', 'credit_score', 'months_employed',
              'num_credit_lines', 'interest_rate', 'loan_term', 'dti_ratio', 'education',
              'employment_type', 'marital_status', 'has_mortgage', 'has_dependents',
              'loan_purpose', 'has_cosigner', 'default_flag']

#convert Yes/No columns to 1/0
for col in ['has_mortgage', 'has_dependents', 'has_cosigner']:
    df[col] = df[col].map({'Yes': 1, 'No': 0})

df.head()
```

```python
#create column age_group
df['age_group'] = pd.cut(df['age'], bins=[17, 29, 39, 49, 59, 100],
                         labels=['18-29', '30-39', '40-49', '50-59', '60+']).astype(str)


#create column income_group (3 equal-sized groups)
df['income_group'] = pd.qcut(df['income'], q=3, labels=['Low', 'Medium', 'High']).astype(str)

#create column borrower_segment (employment type + income group)
df['borrower_segment'] = df['employment_type'] + ' - ' + df['income_group']


#create column emi (monthly instalment) and emi_to_income
rate = df['interest_rate'] / 1200
df['emi'] = df['loan_amount'] * rate * (1 + rate) ** df['loan_term'] / ((1 + rate) ** df['loan_term'] - 1)
df['emi_to_income'] = (df['emi'] / (df['income'] / 12)* 100)

df[['loan_amount', 'interest_rate', 'loan_term', 'income', 'emi', 'emi_to_income']].head()
```

```python
credit_risk = ((850 - df['credit_score']) / 550 * 100).clip(0, 100)
dti_risk = ((df['dti_ratio'] - 0.1) / 0.8 * 100).clip(0, 100)

#EMI-to-income has extreme values, so cap it at the 99th percentile
emi_cap = df['emi_to_income'].quantile(0.99)
emi_risk = ((df['emi_to_income'] - df['emi_to_income'].min()) / (emi_cap - df['emi_to_income'].min()) * 100).clip(0, 100)

df['risk_score'] = (0.35 * credit_risk + 0.35 * dti_risk + 0.30 * emi_risk).round(2)

#create column risk_tier: Low (below 33), Medium (33 to 66), High (66 and above)
df['risk_tier'] = np.select([df['risk_score'] < 33, df['risk_score'] < 66],
                            ['Low', 'Medium'], default='High')

#round emi columns after the score is calculated
df['emi'] = df['emi'].round(2)
df['emi_to_income'] = df['emi_to_income'].round(4)

df[['credit_score', 'dti_ratio', 'emi_to_income', 'risk_score', 'risk_tier']].head(10)
```

```python
credit_risk = ((850 - df['credit_score']) / 550 * 100).clip(0, 100)
dti_risk = ((df['dti_ratio'] - 0.1) / 0.8 * 100).clip(0, 100)

#EMI-to-income has extreme values, so cap it at the 99th percentile
emi_cap = df['emi_to_income'].quantile(0.99)
emi_risk = ((df['emi_to_income'] - df['emi_to_income'].min()) / (emi_cap - df['emi_to_income'].min()) * 100).clip(0, 100)

df['risk_score'] = (0.35 * credit_risk + 0.35 * dti_risk + 0.30 * emi_risk).round(2)

#create column risk_tier: Low (below 33), Medium (33 to 66), High (66 and above)
df['risk_tier'] = np.select([df['risk_score'] < 33, df['risk_score'] < 66],
                            ['Low', 'Medium'], default='High')

#round emi columns after the score is calculated
df['emi'] = df['emi'].round(2)
df['emi_to_income'] = df['emi_to_income'].round(4)

df[['credit_score', 'dti_ratio', 'emi_to_income', 'risk_score', 'risk_tier']].head(10)
```

```python
print(df['risk_tier'].value_counts())
print((df.groupby('risk_tier')['default_flag'].mean() * 100).round(2))

#silent risk: borrowers with credit score 700+ who still defaulted
silent = df[(df['credit_score'] >= 700) & (df['default_flag'] == 1)]
print(len(silent), round(len(silent) / df['default_flag'].sum() * 100, 2))
```
